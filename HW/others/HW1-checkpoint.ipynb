```python
import numpy as np
import sympy as sy
import matplotlib.pyplot as plt

t = sy.symbols('t',real = True)

Wn = 36
damping_ratio = 0.008
T = 0.5
m = 500
Wr0 = 2*sy.pi/T * 0
Wr1 = 2*sy.pi/T * 1
Wr2 = 2*sy.pi/T * 2
Wr3 = 2*sy.pi/T * 3
Wr4 = 2*sy.pi/T * 4


F0 = 0
F1 = 3 + 1j * 2.1
F2 = -0.6 + 1j * 0.5
F3 = 0.84 - 1j * 0.75
F4 = 0.01 - 1j * 0.03

X0 = (F0/m)/((Wn**2)-(Wr0**2) + (1j * 2 * damping_ratio * Wn * Wr0))
X1 = (F1/m)/((Wn**2)-(Wr1**2) + (1j * 2 * damping_ratio * Wn * Wr1))
X2 = (F2/m)/((Wn**2)-(Wr2**2) + (1j * 2 * damping_ratio * Wn * Wr2))
X3 = (F3/m)/((Wn**2)-(Wr3**2) + (1j * 2 * damping_ratio * Wn * Wr3))
X4 = (F4/m)/((Wn**2)-(Wr4**2) + (1j * 2 * damping_ratio * Wn * Wr4))

x_expr = X0 * sy.exp(1j * Wr0 * t) + \
         X1 * sy.exp(1j * Wr1 * t) + \
         X2 * sy.exp(1j * Wr2 * t) + \
         X3 * sy.exp(1j * Wr3 * t) + \
         X4 * sy.exp(1j * Wr4 * t) + \
         sy.conjugate(X0) * sy.exp(-1j * Wr0 * t) + \
         sy.conjugate(X1) * sy.exp(-1j * Wr1 * t) + \
         sy.conjugate(X2) * sy.exp(-1j * Wr2 * t) + \
         sy.conjugate(X3) * sy.exp(-1j * Wr3 * t) + \
         sy.conjugate(X4) * sy.exp(-1j * Wr4 * t)
print('X =')
x_expr

```

```python
f_expr = F0 * sy.exp(1j * Wr0 * t) + \
    F1 * sy.exp(1j * Wr1 * t) +\
    F2 * sy.exp(1j * Wr2 * t) +\
    F3 * sy.exp(1j * Wr3 * t) +\
    F4 * sy.exp(1j * Wr4 * t) +\
sy.conjugate(F0) * sy.exp(-1j * Wr0 * t) + \
sy.conjugate(F1) * sy.exp(-1j * Wr1 * t) + \
sy.conjugate(F2) * sy.exp(-1j * Wr2 * t) + \
sy.conjugate(F3) * sy.exp(-1j * Wr3 * t) + \
sy.conjugate(F4) * sy.exp(-1j * Wr4 * t)
print('f(t)=')
f_expr
```

```python
# 使用 sympy.lambdify 将符号表达式转换为数值计算函数
x_func = sy.lambdify(t, x_expr, 'numpy')
f_func = sy.lambdify(t, f_expr, 'numpy')

# 定义一个函数，输入时间值 t 来计算对应的解
def func(time_values):
    return x_func(time_values)

def func1(time_values):
    return f_func(time_values)

# 生成时间值
time = np.arange(0, 3, 0.001)

# 绘制图像plt.figure()
plt.figure(figsize=(10, 6))
plt.plot(time, np.real(func(time))*100000,"b",label= "x(t)_100(m)")  # 取实部绘制
plt.plot(time, np.real(func1(time)),"r",label="f(t)_(kN)" )
plt.xlabel('Time (s)')
plt.ylabel('Force f(t)')
plt.title('Displacement vs Time')
plt.legend(loc='lower right')
plt.grid(True)
plt.show()
```

```python
# Simplified calculation of Fourier coefficients without error

import sympy as sp

# Define parameters
T = sp.Rational(1,10)  # Period in seconds
omega_0 = 2 * np.pi / T  # Fundamental angular frequency
t,r= sp.symbols("t,r")
Func_e = sp.exp(-sp.I*2*sp.pi*r*t/T)
def f_t(t):
    if 0<= t <T/4:
        return 1
    elif T/4 < t <=3*T/4:
        return -1
    elif 3*T/4<t<=T:
        return 1
cr_list = sp.zeros(15,1)
for r in range(15):
    cr = 1/T * sp.integrate(Func_e.subs({"r":(r-7)}),(t,0,T/4))-1/T * sp.integrate(Func_e.subs({"r":r-7}),(t,T/4,3*T/4)) +1/T * sp.integrate(Func_e.subs({"r":r-7}),(t,3*T/4,T))
    cr_list[r] = cr
```

```python
print("These are the coefficients where r from [-7,7]")
cr_list
```

```python
import matplotlib.pyplot as plt

T=0.1
func = 0
t = np.linspace(0,1, 1000)
for r in range(15):
    func += cr_list[r].evalf()*np.exp(2*np.pi*(r-7)*t/T*1j).real

# Original square wave function over multiple periods
f_original = np.array([f_t(ti % T) for ti in t])
# Plotting
plt.figure(figsize=(10, 6))
plt.plot(t, f_original, label="Original Square Wave", color="magenta")
plt.plot(t, func, label="Fourier Approximation (7 Harmonics)", linestyle="-", color="blue")
plt.title("Square Wave and its Fourier Series Approximation")
plt.xlabel("Time (s)")
plt.ylabel("Amplitude")
plt.legend()
plt.grid()
plt.show()
```
