period $T = 0.5$ s, the fundamental circular frequency is:
$$\omega_0 = \frac{2\pi}{T} = 4\pi \approx 12.57 \text{ rad/s}$$

Transfer Function
$$H(\omega) = \frac{1}{k - m\omega^2 + j(c\omega)}$$
Where:

$k = m\omega_n^2$

$c = 2\zeta m \omega_n$

We calculate the response $x(t)$ for each harmonic separately and sum them up
$$
x(t) = \sum_{n=1}^{4} 2 \cdot \text{Re}(F_n H(n\omega_0) e^{j n \omega_0 t})
$$

```python
import numpy as np
import sympy as sy
import matplotlib.pyplot as plt

t = sy.symbols('t',real = True)

Wn = 36
damping_ratio = 0.008
T = 0.5
m = 500
Wr0 = 2*sy.pi/T * 0
Wr1 = 2*sy.pi/T * 1
Wr2 = 2*sy.pi/T * 2
Wr3 = 2*sy.pi/T * 3
Wr4 = 2*sy.pi/T * 4


F0 = 0
F1 = 3 + 1j * 2.1
F2 = -0.6 + 1j * 0.5
F3 = 0.84 - 1j * 0.75
F4 = 0.01 - 1j * 0.03

X0 = (F0/m)/((Wn**2)-(Wr0**2) + (1j * 2 * damping_ratio * Wn * Wr0))
X1 = (F1/m)/((Wn**2)-(Wr1**2) + (1j * 2 * damping_ratio * Wn * Wr1))
X2 = (F2/m)/((Wn**2)-(Wr2**2) + (1j * 2 * damping_ratio * Wn * Wr2))
X3 = (F3/m)/((Wn**2)-(Wr3**2) + (1j * 2 * damping_ratio * Wn * Wr3))
X4 = (F4/m)/((Wn**2)-(Wr4**2) + (1j * 2 * damping_ratio * Wn * Wr4))

x_expr = X0 * sy.exp(1j * Wr0 * t) + \
         X1 * sy.exp(1j * Wr1 * t) + \
         X2 * sy.exp(1j * Wr2 * t) + \
         X3 * sy.exp(1j * Wr3 * t) + \
         X4 * sy.exp(1j * Wr4 * t) + \
         sy.conjugate(X0) * sy.exp(-1j * Wr0 * t) + \
         sy.conjugate(X1) * sy.exp(-1j * Wr1 * t) + \
         sy.conjugate(X2) * sy.exp(-1j * Wr2 * t) + \
         sy.conjugate(X3) * sy.exp(-1j * Wr3 * t) + \
         sy.conjugate(X4) * sy.exp(-1j * Wr4 * t)
print('X =')
x_expr

```

```python
import numpy as np
import matplotlib.pyplot as plt


m = 500.0           # Mass in kg
wn = 36.0           # Natural freq in rad/s
zeta = 0.008        # Damping ratio
T = 0.5             # Period in seconds


w0 = (2 * np.pi) / T    # Fundamental frequency (rad/s)
k = m * (wn**2)         # Stiffness (N/m)
c = 2 * zeta * m * wn   # Damping coefficient (N*s/m)


F_coeffs_kN = [
    3 + 2.1j,      # F1
    -0.6 + 0.5j,   # F2
    0.84 - 0.75j,  # F3
    0.01 - 0.03j   # F4
]

F_amps = np.array(F_coeffs_kN) * 1000 

t = np.linspace(0, 2*T, 1000)

f_total = np.zeros_like(t)
x_total = np.zeros_like(t)

print(f"System Natural Frequency: {wn:.2f} rad/s")
print("-" * 60)
print(f"{'Harmonic':<10} | {'Freq (rad/s)':<15} | {'Force (N)':<15} | {'Disp (m)':<15}")
print("-" * 60)

for n, Fn in enumerate(F_amps, start=1):
    omega = n * w0
    
    # Calculate Receptance H(omega)
    # H(w) = 1 / (k - mw^2 + jcw)
    H_w = 1 / (k - m*(omega**2) + 1j*c*omega)
    
    # Calculate Response Coefficient Xn
    Xn = Fn * H_w
    
    # --- Time Domain Reconstruction ---
    # We multiply by 2 * Real part to account for both positive 
    # and negative harmonics (rendering a real result).
    
    # Force component
    f_n_t = 2 * np.real(Fn * np.exp(1j * omega * t))
    f_total += f_n_t
    
    # Displacement component
    x_n_t = 2 * np.real(Xn * np.exp(1j * omega * t))
    x_total += x_n_t
    
    print(f"{n:<10} | {omega:<15.2f} | {np.abs(Fn):<15.2f} | {np.abs(Xn):<15.4f}")

plt.figure(figsize=(10, 6))

plt.plot(t, f_total / 1000, 'b', label='Excitation Force $f(t)$ [kN]', linewidth=1.5)

plt.plot(t, x_total * 100, 'r--', label='Response $100 \\times x(t)$ [m]', linewidth=2)

plt.title('Steady State Response vs Excitation Force')
plt.xlabel('Time [s]')
plt.ylabel('Amplitude')
plt.grid(True, which='both', linestyle='--', alpha=0.7)
plt.legend(loc='upper right')
plt.tight_layout()

plt.show()
```
